```python
import numpy as np
```

```python
array_1d = np.arange(1,11)
print(array_1d)
```

```python
array_2d = np.arange(1,10).reshape(3,3)
print(array_2d)
```

```python
np_array = np.array([10,20,30,40,50])
print(np_array)
```

```python
np_array_list = [array_1d, array_2d, np_array]

for arr in np_array_list:
    print(f"Array:\n{arr}\nShape: {arr.shape}\nData Type: {arr.dtype}\n")
```

```python
a = np.array([10.20,30,40,50])
b = np.array([1,2,3,4])
print(f"Addition: {a + b}")
print(f"Subtraction: {a - b}")
print(f"Multiplication: {a * b}")
print(f"Division: {a / b}")
print(f"Exponentiation: {a ** b}")
```

```python
values = np.array([2, 4, 6, 8, 10])

print(f"Square Root: {np.sqrt(values)}")
print(f"Exponential: {np.exp(values)}")
print(f"Natural Logarithm: {np.log(values)}")
print(f"Sum of Elements: {np.sum(values)}")
print(f"Cumulative Sum: {np.cumsum(values)}")
```

```python
data = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print(f"Row wise Sum: {np.sum(data, axis=1)}")
print(f"Column wise Sum: {np.sum(data, axis=0)}")
print(f"Minimum Value: {np.min(data)}")
print(f"Maximum Value: {np.max(data)}")
print(f"Mean Value: {np.mean(data)}")
```

```python
marks = np.array([78,85,90,66,72,88,95,60])
print(f"Mean: {np.mean(marks)}")
print(f"Median: {np.median(marks)}")
print(f"Variance: {np.var(marks)}")
print(f"Standard Deviation: {np.std(marks)}")
print(f"Minimum: {np.min(marks)}")
print(f"Maximum: {np.max(marks)}")
print(f"Range: {np.ptp(marks)}")
# sort the marks array
marks_sorted = np.sort(marks)
print(f"Sorted Marks: {marks_sorted}")
print(f"25th Percentile: {np.percentile(marks, 25)}")
print(f"50th Percentile (Median): {np.percentile(marks, 50)}")
print(f"75th Percentile: {np.percentile(marks, 75)}")
# how many students scored above 50 percentile
above_50_percentile = marks[marks > np.percentile(marks, 50)]
print(f"Number of Students above 50th Percentile: {len(above_50_percentile)}")
```

```python
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])
print(f"Total weekly sales: {np.sum(sales)}")
print(f"Mean weekly sales: {np.mean(sales)}")
print(f"Maximum weekly sales: {np.max(sales)}")
print(f"Minimum weekly sales: {np.min(sales)}")
print(f"Standard Deviation of weekly sales: {np.std(sales)}")
print(f"Sales above average: {sales[sales > np.mean(sales)]}")
```

# Task 1: Basic Function - Price After Discount

Write a function `apply_discount(price, discount_percent)` that:

1. Returns the price after applying a discount.
2. Uses a default discount of 5% when `discount_percent` is not provided.
3. Tests `apply_discount(1000, 10)` and `apply_discount(500)`.
4. Ensures the discount does not exceed 60%.

```python
def apply_discount(price, discount_percent=5):
    if discount_percent < 0:
        raise ValueError("Discount percent cannot be negative.")
    elif discount_percent > 60:
        print("Warning: Discount percent cannot exceed 60%. Applying maximum discount of 60%.")
    discount_percent = min(discount_percent, 60)
    return price * (1 - discount_percent / 100)

print("Price after 10% discount:", apply_discount(1000, 10))
print("Price after default 5% discount:", apply_discount(500))
print("Price after 70% discount (should apply max 60%):", apply_discount(200, 70))
```

# Task 2: Recursive Function - Factorial Utility

Create a recursive function `factorial(n)` that:

1. Returns the factorial of `n`.
2. Handles the edge cases `n == 0` and `n == 1`.
3. Prints an error message when `n` is negative.
4. Tests `factorial(5)`, `factorial(0)`, and `factorial(-3)`.

```python
def factorial(n):
    if n < 0:
        print("Error: factorial is not defined for negative numbers.")
        return None
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print("factorial(5):", factorial(5))
print("factorial(0):", factorial(0))
print("factorial(-3):", factorial(-3))
```

# Task 3: Lambda Function - GST Calculator

Create a lambda function `gst` that adds 18% GST to a price.

Test it with `gst(100)`, which should return `118`.

As an extra, calculate a final price after applying a discount and then adding GST.

```python
gst = lambda price: price + (0.18 * price)

print("Price after GST:", gst(100))

# Apply the discount first, then add 18% GST.
final_price = lambda price, discount: gst(price * (1 - discount / 100))
print("Price after 10% discount and GST:", final_price(100, 10))
```

# Task 4: Using `map()` - Apply GST to a List of Prices

Given `prices = [100, 250, 400, 1200, 50]`, use `map()` with the `gst` lambda to create `prices_with_gst`.

Print the original prices and the prices after GST.

# Task 5: Using `filter()` - Filter Expensive Products

Given `prices = [100, 250, 400, 1200, 50, 2000, 850]`:

1. Use `filter()` to create a list of prices greater than 500.
2. Use `filter()` to create another list of prices less than or equal to 500.
3. Print both lists.

```python
prices = [100, 250, 400, 1200, 50, 2000, 850]

expensive_prices = list(filter(lambda price: price > 500, prices))
affordable_prices = list(filter(lambda price: price <= 500, prices))

print("Prices greater than 500:", expensive_prices)
print("Prices less than or equal to 500:", affordable_prices)
```

# Task 6: Combined Utility Function

Write a function `process_prices(prices)` that:

1. Uses `map()` and a lambda to apply a 10% discount to every price.
2. Uses `filter()` to keep only discounted prices above 300.
3. Returns both `discounted_prices` and `filtered_prices`.

Test the function with `[100, 500, 900, 50, 750]`.

```python
def process_prices(prices):
    discounted_prices = list(map(lambda price: price * 0.90, prices))
    filtered_prices = list(filter(lambda price: price > 300, discounted_prices))
    return discounted_prices, filtered_prices

prices = [100, 500, 900, 50, 750]
discounted_prices, filtered_prices = process_prices(prices)

print("Discounted prices:", discounted_prices)
print("Discounted prices above 300:", filtered_prices)
```

# Task 7: Mini Problem - Menu Using Functions

Create three functions:

1. `add_price(prices_list, price)` adds a price to the list.
2. `get_average_price(prices_list)` returns the average price.
3. `get_max_price(prices_list)` returns the maximum price.

Use these functions in a loop-driven menu:

- `1` - Add price
- `2` - Show average price
- `3` - Show highest price
- `q` - Quit

```python
def add_price(prices_list, price):
    prices_list.append(price)


def get_average_price(prices_list):
    if not prices_list:
        return 0
    return sum(prices_list) / len(prices_list)


def get_max_price(prices_list):
    if not prices_list:
        return None
    return max(prices_list)


prices_list = []

while True:
    print("\n1 - Add price")
    print("2 - Show average price")
    print("3 - Show highest price")
    print("q - Quit")
    choice = input("Choose an option: ").strip().lower()

    if choice == "1":
        try:
            price = float(input("Enter a price: "))
            add_price(prices_list, price)
            print("Price added.")
        except ValueError:
            print("Please enter a valid number.")
    elif choice == "2":
        if prices_list:
            print("Average price:", get_average_price(prices_list))
        else:
            print("No prices have been added yet.")
    elif choice == "3":
        if prices_list:
            print("Highest price:", get_max_price(prices_list))
        else:
            print("No prices have been added yet.")
    elif choice == "q":
        print("Goodbye!")
        break
    else:
        print("Invalid option.")
```

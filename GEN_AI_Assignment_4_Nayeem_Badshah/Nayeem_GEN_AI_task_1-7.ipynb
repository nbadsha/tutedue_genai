# Task 1: Write Sales Records to a File

Create a list of sales amounts and write each sale on a new line in `sales_data.txt`.

Then close the file, reopen it, and print its contents.

```python
sales = [1200, 450, 980, 1500, 3000]

with open("sales_data.txt", "w") as file:
    for sale in sales:
        file.write(f"{sale}\n")

with open("sales_data.txt", "r") as file:
    contents = file.read()

print("Sales file contents:")
print(contents)
```

# Task 2: Read File in Different Ways

Using `sales_data.txt`:

1. Read and print the entire file with `.read()`.
2. Read and print the first line with `.readline()`.
3. Read all lines with `.readlines()`, remove newline characters, and convert the values to integers.

```python
with open("sales_data.txt", "r") as file:
    entire_file = file.read()
print("Entire file:")
print(entire_file)

with open("sales_data.txt", "r") as file:
    first_line = file.readline().strip()
print("First line:")
print(first_line)

with open("sales_data.txt", "r") as file:
    lines = file.readlines()

sales_as_integers = [int(line.strip()) for line in lines]
print("Sales as integers:")
print(sales_as_integers)
```

# Task 3: Append New Sales

Append the new sales `5000`, `2500`, and `1700` to `sales_data.txt`.

Then reopen the file and print the entire updated file. Also print the total number of lines.

```python
new_sales = [5000, 2500, 1700]

with open("sales_data.txt", "a") as file:
    for sale in new_sales:
        file.write(f"{sale}\n")

with open("sales_data.txt", "r") as file:
    updated_lines = file.readlines()

print("Updated sales file:")
print("".join(updated_lines))
print("Total number of lines:", len(updated_lines))
```

# Task 4: Generate Summary Report from File

Read all sales values from `sales_data.txt`, convert them to integers, and calculate:

- Total Sales
- Highest Sale
- Lowest Sale
- Average Sale

Use only basic Python file-reading operations and built-in calculations.

```python
with open("sales_data.txt", "r") as file:
    sales_values = [int(line.strip()) for line in file.readlines()]

total_sales = sum(sales_values)
highest_sale = max(sales_values)
lowest_sale = min(sales_values)
average_sale = total_sales / len(sales_values)

print("Total Sales:", total_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
print("Average Sale:", average_sale)
```

# Task 5: Create Product Info File

Ask the user for three product names and their prices.

Write the information to `products.txt` in the format `ProductName | Price`, then read the file and print each line with proper formatting.

```python
products = []

for number in range(3):
    product_name = input(f"Enter product {number + 1} name: ").strip()
    price = float(input(f"Enter price for {product_name}: "))
    products.append((product_name, price))

with open("products.txt", "w") as file:
    for product_name, price in products:
        file.write(f"{product_name} | {price:.2f}\n")

with open("products.txt", "r") as file:
    print("Product information:")
    for line in file:
        print(line.strip())
```

# Task 6: Read File Safely

Ask the user for a filename. Use `os.path.exists()` to check whether the file exists before opening it.

- If the file exists, read and print its contents.
- Otherwise, print: `File not found. Please check the filename.`

```python
import os

filename = input("Enter the filename to open: ").strip()

if os.path.exists(filename):
    with open(filename, "r") as file:
        print(file.read())
else:
    print("File not found. Please check the filename.")
```

# Task 7: Mini Project - Export Discounted Prices

Create a product-price dictionary and ask the user for a discount percentage.

Write a discount report to `discount_report.txt` with the columns `Product | Original Price | Discounted Price`. Add the total item count and average discounted price, then read and print the report.

```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000,
}

discount_percent = float(input("Enter discount percentage: "))
discount_report = []

for product, original_price in prices.items():
    discounted_price = original_price * (1 - discount_percent / 100)
    discount_report.append((product, original_price, discounted_price))

with open("discount_report.txt", "w") as file:
    file.write("Product | Original Price | Discounted Price\n")
    for product, original_price, discounted_price in discount_report:
        file.write(f"{product} | {original_price:.2f} | {discounted_price:.2f}\n")

    average_discounted_price = sum(item[2] for item in discount_report) / len(discount_report)
    file.write(f"\nTotal Items: {len(discount_report)}\n")
    file.write(f"Average Discounted Price: {average_discounted_price:.2f}\n")

with open("discount_report.txt", "r") as file:
    print(file.read())
```

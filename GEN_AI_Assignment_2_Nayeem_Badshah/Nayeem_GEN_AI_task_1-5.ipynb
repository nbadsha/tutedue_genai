# Task 1: Discount Rules

Write a program that:

1. Reads an integer `order_amount` from the user using `input()`.
2. Applies these discount rules:
   - `order_amount >= 2000`: 15% discount
   - `1500 <= order_amount < 2000`: 10% discount
   - `1000 <= order_amount < 1500`: 7% discount
   - Otherwise: 0% discount
3. Prints the final amount after applying the discount.
4. Converts the input to a number and displays an error message when the input is non-numeric.

```python
try:
    # Read and convert the order amount to an integer.
    order_amount = int(input("Enter the order amount: "))
except ValueError:
    print("Error: order amount must be a whole number.")
else:
    # Select the discount rate based on the order amount.
    if order_amount >= 2000:
        discount_rate = 0.15
    elif order_amount >= 1500:
        discount_rate = 0.10
    elif order_amount >= 1000:
        discount_rate = 0.07
    else:
        discount_rate = 0.0

    discount_amount = order_amount * discount_rate
    final_amount = order_amount - discount_amount

    print(f"Order amount: {order_amount:.2f}")
    print(f"Discount: {discount_rate:.0%}")
    print(f"Final amount: {final_amount:.2f}")
```

# Task 2: Process Multiple Orders

Given `orders = [1200, 2500, 800, 1750, 3000]`, use a `for` loop to:

1. Apply the discount rules from Task 1 to each order.
2. Print a summary table showing `order_amount -> discount% -> final_amount`.
3. Calculate and print the total revenue after discounts.
4. Print the number of orders that received a discount.

```python
orders = [1200, 2500, 800, 1750, 3000]
total_revenue = 0
orders_with_discount = 0

print("Order Amount -> Discount% -> Final Amount")
print("------------------------------------------")

for order_amount in orders:
    # Apply the appropriate discount tier to this order.
    if order_amount >= 2000:
        discount_rate = 0.15
    elif order_amount >= 1500:
        discount_rate = 0.10
    elif order_amount >= 1000:
        discount_rate = 0.07
    else:
        discount_rate = 0.0

    final_amount = order_amount * (1 - discount_rate)
    total_revenue += final_amount

    # Count orders that received a non-zero discount.
    if discount_rate > 0:
        orders_with_discount += 1

    print(f"{order_amount:12.2f} -> {discount_rate:8.0%} -> {final_amount:12.2f}")

print(f"\nTotal revenue after discounts: {total_revenue:.2f}")
print(f"Orders that received a discount: {orders_with_discount}")
```

# Task 3: User Menu

Implement a `while`-loop driven menu that continues until the user enters `q`:

1. Add an order amount to a running list using `input()`.
2. Show all orders and their totals after applying the discount rules.
3. Use `continue` to re-show the menu after invalid input.
4. Use `break` to exit when the user enters `q`.

Keep the loop logic simple and use only lists and loops.

```python
orders = []

while True:
    print("\n1 - Add order amount")
    print("2 - Show orders and totals")
    print("q - Quit")
    choice = input("Choose an action: ").strip().lower()

    if choice == "1":
        try:
            # Convert the entered amount before adding it to the list.
            order_amount = int(input("Enter the order amount: "))
            orders.append(order_amount)
            print("Order added.")
        except ValueError:
            print("Error: order amount must be a whole number.")
        continue

    if choice == "2":
        total_revenue = 0
        print("\nOrder Amount -> Discount% -> Final Amount")
        print("------------------------------------------")

        for order_amount in orders:
            # Apply the same discount tiers used in the earlier tasks.
            if order_amount >= 2000:
                discount_rate = 0.15
            elif order_amount >= 1500:
                discount_rate = 0.10
            elif order_amount >= 1000:
                discount_rate = 0.07
            else:
                discount_rate = 0.0

            final_amount = order_amount * (1 - discount_rate)
            total_revenue += final_amount
            print(f"{order_amount:12.2f} -> {discount_rate:8.0%} -> {final_amount:12.2f}")

        print(f"Total revenue after discounts: {total_revenue:.2f}")
        continue

    if choice == "q":
        print("Goodbye!")
        break

    print("Invalid choice. Please select 1, 2, or q.")
    continue
```

# Task 4: Loop Control with Conditions

Given `daily = [200, 150, 0, 400, 50, -1, 300]`, use a `for` loop to process daily sales:

1. Treat `-1` as corrupted data and use `break` to stop processing.
2. Treat `0` as a day with no sales and use `continue` to skip it.
3. Add valid positive sales to `total_sales` and print the running total.
4. Print the final total after the loop completes or stops.

```python
daily = [200, 150, 0, 400, 50, -1, 300]
total_sales = 0

for sales in daily:
    # Stop processing when corrupted data is found.
    if sales == -1:
        print("Corrupted data found. Stopping processing.")
        break

    # Skip days with no sales.
    if sales == 0:
        continue

    # Add valid positive sales to the running total.
    total_sales += sales
    print(f"Running total: {total_sales}")

print(f"Final total sales: {total_sales}")
```
